# Exploratory Data Analysis: Nigerian Monetary Policy Transmission

**Objective**: Visualize time series properties of key macroeconomic variables

**Variables**:
- Monetary Policy Rate (MPR)
- Headline Inflation (CPI YoY %)
- Exchange Rate (NGN/USD)
- Money Supply (M2)

**Period**: 2010-01 to 2025-01 (181 observations)

```python
# Import libraries
import sys
sys.path.append('../src')

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Configure plotting
plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette("husl")
%matplotlib inline

# Figure size default
plt.rcParams['figure.figsize'] = (14, 6)
plt.rcParams['font.size'] = 11

print("✓ Libraries loaded successfully")
```

## 1. Load Data

```python
from data_ingestion.data_loader import NigerianMacroDataLoader

# Initialize loader
loader = NigerianMacroDataLoader(data_dir='../data')

# Load and prepare data
df = loader.load_and_prepare()
```

```python
# Display first and last 10 rows
print("First 10 observations:")
display(df.head(10))

print("\nLast 10 observations:")
display(df.tail(10))
```

## 2. Descriptive Statistics

```python
# Summary statistics
print("=" * 70)
print("DESCRIPTIVE STATISTICS (2010-2025)")
print("=" * 70)
display(df.describe().T)
```

```python
# Correlation matrix
print("\nCorrelation Matrix:")
corr_matrix = df.corr()
display(corr_matrix)

# Visualize correlation heatmap
plt.figure(figsize=(8, 6))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', center=0, 
            square=True, linewidths=1, cbar_kws={"shrink": 0.8})
plt.title('Correlation Matrix: Nigerian Macroeconomic Variables', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

## 3. Time Series Visualization

```python
# Plot all variables in separate subplots
fig, axes = plt.subplots(4, 1, figsize=(14, 12))

# MPR
axes[0].plot(df.index, df['MPR'], color='#2E86AB', linewidth=2)
axes[0].set_title('Monetary Policy Rate (MPR)', fontsize=12, fontweight='bold')
axes[0].set_ylabel('% per annum')
axes[0].grid(True, alpha=0.3)
axes[0].axhline(y=df['MPR'].mean(), color='red', linestyle='--', alpha=0.5, label='Mean')
axes[0].legend()

# Inflation
axes[1].plot(df.index, df['Inflation'], color='#A23B72', linewidth=2)
axes[1].set_title('Headline Inflation (YoY)', fontsize=12, fontweight='bold')
axes[1].set_ylabel('% YoY')
axes[1].grid(True, alpha=0.3)
axes[1].axhline(y=df['Inflation'].mean(), color='red', linestyle='--', alpha=0.5, label='Mean')
axes[1].legend()

# Exchange Rate
axes[2].plot(df.index, df['ExchangeRate'], color='#F18F01', linewidth=2)
axes[2].set_title('Exchange Rate (NGN/USD)', fontsize=12, fontweight='bold')
axes[2].set_ylabel('Naira per Dollar')
axes[2].grid(True, alpha=0.3)

# M2
axes[3].plot(df.index, df['M2'], color='#6A994E', linewidth=2)
axes[3].set_title('Broad Money Supply (M2)', fontsize=12, fontweight='bold')
axes[3].set_ylabel('Naira Billions')
axes[3].set_xlabel('Date')
axes[3].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("✓ Time series plots generated")
```

## 4. Key Economic Episodes

**Structural breaks to investigate**:
- **2016**: Major Naira devaluation (₦197 → ₦305)
- **2020**: COVID-19 shock and CBN policy response
- **2023**: FX market unification and inflation surge

```python
# Highlight key episodes
fig, ax = plt.subplots(figsize=(14, 6))

# Plot inflation and MPR on same chart
ax.plot(df.index, df['MPR'], label='MPR', color='#2E86AB', linewidth=2)
ax.plot(df.index, df['Inflation'], label='Inflation', color='#A23B72', linewidth=2)

# Mark key episodes
ax.axvline(pd.Timestamp('2016-06-01'), color='red', linestyle='--', alpha=0.7, label='2016 Devaluation')
ax.axvline(pd.Timestamp('2020-03-01'), color='orange', linestyle='--', alpha=0.7, label='COVID-19')
ax.axvline(pd.Timestamp('2023-06-01'), color='purple', linestyle='--', alpha=0.7, label='2023 FX Unification')

ax.set_title('MPR vs Inflation: Key Economic Episodes', fontsize=14, fontweight='bold')
ax.set_ylabel('Percent (%)')
ax.set_xlabel('Date')
ax.legend(loc='upper left')
ax.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## 5. Growth Rates (First Differences)

```python
# Calculate first differences
df_diff = df.diff().dropna()

fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# MPR changes
axes[0, 0].plot(df_diff.index, df_diff['MPR'], color='#2E86AB', linewidth=1.5)
axes[0, 0].axhline(y=0, color='black', linestyle='-', alpha=0.3)
axes[0, 0].set_title('MPR Changes (First Difference)', fontweight='bold')
axes[0, 0].set_ylabel('Percentage Points')
axes[0, 0].grid(True, alpha=0.3)

# Inflation changes
axes[0, 1].plot(df_diff.index, df_diff['Inflation'], color='#A23B72', linewidth=1.5)
axes[0, 1].axhline(y=0, color='black', linestyle='-', alpha=0.3)
axes[0, 1].set_title('Inflation Changes', fontweight='bold')
axes[0, 1].set_ylabel('Percentage Points')
axes[0, 1].grid(True, alpha=0.3)

# Exchange rate changes
axes[1, 0].plot(df_diff.index, df_diff['ExchangeRate'], color='#F18F01', linewidth=1.5)
axes[1, 0].axhline(y=0, color='black', linestyle='-', alpha=0.3)
axes[1, 0].set_title('Exchange Rate Changes', fontweight='bold')
axes[1, 0].set_ylabel('NGN/USD')
axes[1, 0].grid(True, alpha=0.3)

# M2 changes
axes[1, 1].plot(df_diff.index, df_diff['M2'], color='#6A994E', linewidth=1.5)
axes[1, 1].axhline(y=0, color='black', linestyle='-', alpha=0.3)
axes[1, 1].set_title('M2 Changes', fontweight='bold')
axes[1, 1].set_ylabel('Naira Billions')
axes[1, 1].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()
```

## 6. Distribution Analysis

```python
# Histograms
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

axes[0, 0].hist(df['MPR'], bins=20, color='#2E86AB', edgecolor='black', alpha=0.7)
axes[0, 0].set_title('MPR Distribution', fontweight='bold')
axes[0, 0].set_xlabel('% per annum')
axes[0, 0].set_ylabel('Frequency')

axes[0, 1].hist(df['Inflation'], bins=20, color='#A23B72', edgecolor='black', alpha=0.7)
axes[0, 1].set_title('Inflation Distribution', fontweight='bold')
axes[0, 1].set_xlabel('% YoY')
axes[0, 1].set_ylabel('Frequency')

axes[1, 0].hist(df['ExchangeRate'], bins=20, color='#F18F01', edgecolor='black', alpha=0.7)
axes[1, 0].set_title('Exchange Rate Distribution', fontweight='bold')
axes[1, 0].set_xlabel('NGN/USD')
axes[1, 0].set_ylabel('Frequency')

axes[1, 1].hist(df['M2'], bins=20, color='#6A994E', edgecolor='black', alpha=0.7)
axes[1, 1].set_title('M2 Distribution', fontweight='bold')
axes[1, 1].set_xlabel('Naira Billions')
axes[1, 1].set_ylabel('Frequency')

plt.tight_layout()
plt.show()
```

**Key Takeaways** (to be validated in formal tests):
